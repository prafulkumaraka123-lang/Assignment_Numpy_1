## Question 1 – File modes and reading methods

### Answer
Python `open()` modes include `r` (read; file must exist), `w` (write; creates or truncates), `a` (append; creates if absent), `x` (exclusive creation; fails if present), `r+` (read/write; must exist), `w+` (read/write; truncates or creates), and `a+` (read/append; creates if absent). Binary versions include `rb`, `wb`, and `ab`.

`.read()` reads the whole file or a requested amount; `.readline()` reads one line; `.readlines()` reads all lines into a list; iterating directly over a file reads incrementally and is generally most memory-efficient for very large files.


```python
# Q1
with open("sample.txt", "w") as f:
    f.write("First line\nSecond line\nThird line\n")

with open("sample.txt", "r") as f:
    print("Using read():")
    print(f.read())

with open("sample.txt", "r") as f:
    print("Using direct iteration:")
    for line in f:
        print(line.strip())
```

## Question 2 – `w`, `a`, closing files and `with`

### Answer
`w` mode truncates existing content, so it can accidentally erase data. `a` preserves existing content and adds data at the end. Files should be closed to release resources and flush buffered data. A manual `f.close()` may be skipped if an exception occurs first. The `with` statement automatically closes the file even when an exception occurs.


```python
# Q2
lines = ["Python\n", "File handling\n", "Assignment work\n"]
with open("notes.txt", "w") as f:
    f.writelines(lines)
with open("notes.txt", "a") as f:
    f.write("This line was appended.\n")
with open("notes.txt", "r") as f:
    print(f.read())
```

## Question 4 – Errors and try/except/else/finally

### Answer
A syntax error violates Python grammar and prevents execution. A runtime error/exception occurs during execution, such as `10 / 0`. A logical error lets the program run but produces an incorrect result.

For `divide(10, 2)`: `Division successful: 5.0` followed by `Divide operation attempted.` For `divide(10, 0)`: `Cannot divide by zero!` followed by `Divide operation attempted.` The `else` block runs only when `try` succeeds; `finally` runs whether an exception occurs or not.


```python
# Q4
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero!")
    else:
        print("Division successful:", result)
    finally:
        print("Divide operation attempted.")

divide(10, 2)
divide(10, 0)
```

## Question 5 – Custom exception

### Answer
A bare `except:` is discouraged because it catches almost every exception and can hide unexpected problems. Catch the most specific expected exception. The custom `InsufficientFundsError` below is raised when the withdrawal exceeds the balance.


```python
# Q5
class InsufficientFundsError(Exception):
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(f"Insufficient funds: balance is ₹{balance}, requested ₹{amount}.")
    return balance - amount

try:
    print("Remaining balance:", withdraw(5000, 1500))
except InsufficientFundsError as e:
    print("Withdrawal failed:", e)

try:
    print("Remaining balance:", withdraw(5000, 6000))
except InsufficientFundsError as e:
    print("Withdrawal failed:", e)
```

## Question 6 – Logging

### Answer
Logging provides severity levels, timestamps/metadata, file output, and centralized configuration. Increasing severity is `DEBUG`, `INFO`, `WARNING`, `ERROR`, `CRITICAL`. Typical uses: DEBUG for diagnostics, INFO for normal progress, WARNING for unusual but recoverable situations, ERROR for failed operations, and CRITICAL for severe failures.


```python
# Q6
import logging
logging.basicConfig(filename="app.log", level=logging.DEBUG, format="%(asctime)s - %(levelname)s - %(message)s", force=True)
logging.debug("Debug information.")
logging.info("Application started successfully.")
logging.warning("A configuration value is missing.")
logging.error("An operation failed.")
print("Log messages written to app.log.")
print(open("app.log").read())
```

## Question 7 – `logging.error()` versus `logging.exception()`

### Answer
When the missing file is opened, `FileNotFoundError` is caught. `logging.exception("Failed to load config file")` writes an ERROR-level record containing the message and the active exception traceback. `logging.error()` records an error message but does not automatically include that traceback. `logging.exception()` should be used inside an `except` block.


```python
# Q7
import logging
logging.basicConfig(filename="errors.log", level=logging.DEBUG, format="%(asctime)s - %(levelname)s - %(message)s", force=True)

def load_config(path):
    try:
        with open(path) as f:
            return f.read()
    except FileNotFoundError:
        logging.exception("Failed to load config file")
        return None

print("Returned value:", load_config("does_not_exist.cfg"))
print("\nerrors.log:")
print(open("errors.log").read())
```

## Question 8 – NumPy array creation and attributes

### Answer
Arrays can be created from a Python list with `np.array()`, or with `np.zeros()`, `np.ones()`, `np.full()`, `np.eye()`, `np.arange()`, and `np.linspace()`. These are useful respectively for converting existing data, initialization, constant-filled matrices, identity matrices, sequences/indices, and evenly spaced numerical samples.

`shape` gives dimensions; `ndim` gives number of dimensions; `size` gives total elements; `dtype` gives element type; `itemsize` gives bytes per element.


```python
# Q8
import numpy as np
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
print("shape:", arr.shape)
print("ndim:", arr.ndim)
print("size:", arr.size)
print("dtype:", arr.dtype)
print("itemsize:", arr.itemsize)
print("\nzeros:", np.zeros((2, 3)))
print("\nones:", np.ones((2, 3)))
print("\nfull:", np.full((2, 3), 7))
print("\neye:", np.eye(3))
print("\narange:", np.arange(0, 10, 2))
print("\nlinspace:", np.linspace(0, 1, 5))
```

## Question 9 – NumPy slicing, boolean indexing and fancy indexing

### Answer
Basic slicing generally returns a **view**, so modifying it can modify the original array. Boolean indexing and fancy integer-array indexing generally return **copies**. `.copy()` explicitly creates an independent copy. This distinction matters for large datasets because views can avoid memory-heavy copies, while copies provide independent data when needed.


```python
# Q9
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
view = arr[1:4]
view[0] = 999
print("Original after slice modification:", arr)
copy_arr = arr[1:4].copy()
copy_arr[0] = 111
print("Original after copy modification:", arr)
print("Independent copy:", copy_arr)
selected = arr[arr > 30]
selected[0] = -1
print("Boolean result:", selected)
print("Original:", arr)
fancy = arr[[0, 2, 4]]
fancy[0] = -2
print("Fancy result:", fancy)
print("Original:", arr)
```

## Question 10 – Broadcasting, axes, reshape and stacking

### Answer
Broadcasting lets NumPy operate on compatible arrays without manually copying the smaller array. For a 2D array, `np.sum(arr, axis=0)` produces column sums, while `axis=1` produces row sums. `reshape()` changes shape, `flatten()` creates a 1D copy, `np.hstack()` stacks horizontally, and `np.vstack()` stacks vertically.


```python
# Q10
import numpy as np
arr = np.array([[1, 2, 3], [4, 5, 6]])
values = np.array([10, 20, 30])
print("Broadcasting:")
print(arr + values)
print("\naxis=0 column sums:")
print(np.sum(arr, axis=0))
print("\naxis=1 row sums:")
print(np.sum(arr, axis=1))
one_d = np.array([1, 2, 3, 4, 5, 6])
two_d = one_d.reshape(2, 3)
print("\nReshaped 2x3:")
print(two_d)
print("\nFlattened:")
print(two_d.flatten())
print("\nHorizontal stack:")
print(np.hstack((np.array([[1], [2]]), np.array([[3], [4]]))))
another = np.array([[7, 8, 9], [10, 11, 12]])
print("\nVertical stack:")
print(np.vstack((two_d, another)))
```
